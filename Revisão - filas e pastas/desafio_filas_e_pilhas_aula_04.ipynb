```python
class editor_de_texto:
  def __init__(self):
    self.texto = ""
    self.historico = [] # Esta lista será a nossa Pilha

  def digitar_texto(self, texto_inserido):
    # 1. Salva o estado atual na pilha ANTES de alterar
    self.historico.append(self.texto)
    # 2. Aplica a nova ação
    self.texto += texto_inserido

  def apagar_texto(self, qtd_caracteres):
    # 1. Salva o estado atual na pilha
    self.historico.append(self.texto)
    # 2. Remove os últimos caracteres (usando fatiamento/slicing)
    self.texto = self.texto[:-qtd_caracteres]

  def desfazer(self):
    # Verifica se há algo no histórico para desfazer
    if len(self.historico) > 0:
      # .pop() remove e retorna o último elemento adicionado à pilha
      self.texto = self.historico.pop()
    else:
      print("Não há ações para desfazer!")

  # 1. Criando o objeto (aqui o __init__ é acionado invisivelmente)
meu_editor = editor_de_texto()

# 2. Testando a digitação
meu_editor.digitar_texto("Olá")
print("Ação 1 - Digitar:", meu_editor.texto)

meu_editor.digitar_texto(" Rodrigo!")
print("Ação 2 - Digitar:", meu_editor.texto)

# 3. Testando o apagar (vamos apagar os últimos 8 caracteres: "Rodrigo!")
meu_editor.apagar_texto(8)
print("Ação 3 - Apagar:", meu_editor.texto)

# 4. Testando a Pilha com o desfazer
meu_editor.desfazer()
print("Ação 4 - Desfazer o apagar:", meu_editor.texto)

meu_editor.desfazer()
print("Ação 5 - Desfazer a 2ª digitação:", meu_editor.texto)

meu_editor.desfazer()
print("Ação 6 - Desfazer a 1ª digitação:", meu_editor.texto)
```

```python
class sistema_de_impressao:
  #1 Construtor (como se fosse o public SistemaDeImpressao() no java)
  def __init__(self):
    self.fila_de_impressao = []

  #2. Método para adicionar na fila
  def adicionar_na_fila(self, documento):
    self.fila_de_impressao.append(documento)
    print(f"Documento {documento} adicionado na fila.")

  # 3. Método para imprimir (remover da fila)
  def imprimir_documento(self):
    if len(self.fila_de_impressao) > 0:
      # O pulo do gato está aqui: o 0 dentro do pop()
      documento_atual = self.fila_de_impressao.pop(0)
      print(f"Imprimindo o documento {documento_atual}")
    else:
      print("A fila está vazia!! Não há nada para imprimir!")

# 1 ligando a impressora (Instanciando o objeto)
minha_impressora = sistema_de_impressao()

print("--- Recebendo Documentos na fila---")

# 2. Mandando documentos para a impressora
minha_impressora.adicionar_na_fila("Curriculo_Rodrigo.PDF")
minha_impressora.adicionar_na_fila("Trabalho_faculdade.DOCX")
minha_impressora.adicionar_na_fila("Boleto_faculdade.pdf")

print("\n--- Iniciando as impressões ---")
# 3. Imprimindo na ordem correta (o mais antigo primeiro)
minha_impressora.imprimir_documento() # deve imprimir o currículo
minha_impressora.imprimir_documento() # deve imprimir o trabalho
minha_impressora.imprimir_documento() # deve imprimir o boleto

# 4. Testando quando a fila já acabou
minha_impressora.imprimir_documento()
```

```python
# 1. Precisamos importar a ferramenta 'deque' que vem no Python
from collections import deque

class triagem_hospitalar:
    def __init__(self):
        # Em vez de criar uma lista normal [], criamos um deque vazio
        self.fila = deque()

    # Método para o fluxo padrão (pacientes normais)
    def adicionar_paciente_normal(self, nome):
        # .append() coloca o paciente no final da fila (à direita)
        self.fila.append(nome)
        print(f"Normal: {nome} entrou no final da fila.")

    # Método para o fluxo de exceção (prioridades)
    def adicionar_paciente_prioritario(self, nome):
        # .appendleft() é a mágica! Coloca o paciente direto no início (à esquerda)
        self.fila.appendleft(nome)
        print(f"PRIORIDADE: {nome} furou a fila e foi para o início!")

    # Método para o médico chamar o próximo
    def chamar_proximo(self):
        # Verifica se tem alguém na fila (len > 0)
        if len(self.fila) > 0:
            # .popleft() remove e devolve o primeiro da fila (o mais à esquerda)
            paciente = self.fila.popleft()
            print(f"Médico chamando: {paciente}")
        else:
            print("Não há pacientes aguardando.")

# 1. Abrindo o hospital (Instanciando o objeto)
meu_hospital = triagem_hospitalar()

print("--- Chegada de Pacientes ---")
# 2. Chegam dois pacientes normais
meu_hospital.adicionar_paciente_normal("Rodrigo (Dor de cabeça)")
meu_hospital.adicionar_paciente_normal("Ana (Exame de rotina)")

# 3. O médico chama o primeiro (Deve ser o Rodrigo)
print("\n--- Atendimento 1 ---")
meu_hospital.chamar_proximo()

# 4. Chega uma emergência no meio do processo!
print("\n--- Chegada de Pacientes ---")
meu_hospital.adicionar_paciente_prioritario("Senhor Paulo (85 anos, falta de ar)")
meu_hospital.adicionar_paciente_normal("Carlos (Tosse)")

# 5. O médico chama os próximos
print("\n--- Atendimentos Seguintes ---")
# Aqui a mágica acontece: O Senhor Paulo DEVE ser o próximo, passando na frente da Ana
meu_hospital.chamar_proximo() # Chama o Senhor Paulo
meu_hospital.chamar_proximo() # Chama a Ana (que estava esperando antes do Carlos)
meu_hospital.chamar_proximo() # Chama o Carlos
```
